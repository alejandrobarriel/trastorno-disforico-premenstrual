# Detecting Early Physiological Deterioration and Optimizing the Timing of Intervention
### Physiological Deviations as a Basis for Intervention Decisions

---
## I. Contexto y Justificación

## 1. El Problema

El síndrome premenstrual (PMS) afecta a entre el 30% y el 80% de las mujeres en edad reproductiva, dependiendo del criterio diagnóstico empleado. Su forma más severa, el trastorno disfórico premenstrual (PMDD), tiene una prevalencia de 3.2% en diagnóstico confirmado y 7.7% en diagnóstico provisional según el meta-análisis más reciente (Reilly et al., 2024). PMDD está clasificado como trastorno depresivo en el DSM-5 desde 2013, y produce afectación funcional comparable a la del trastorno depresivo mayor durante la fase lútea tardía del ciclo.

El problema estructural es el mismo que en otras condiciones recurrentes con expresión fluctuante: **se reconoce tarde**. La evaluación clínica estándar de PMS/PMDD se basa en autoinforme prospectivo durante al menos dos ciclos, mediante instrumentos validados como el DRSP (Daily Record of Severity of Problems) o el COPE (Calendar of Premenstrual Experiences). La revisión sistemática de Tsuyuki et al. (2025) identifica 11 escalas distintas en uso, con heterogeneidad metodológica que limita la comparabilidad entre estudios. La carga de evaluación es alta y el cumplimiento variable.

El problema no es que estos instrumentos sean malos. Como ocurre en depresión con el PHQ-9 o el BDI-II, están diseñados para medir severidad de síntomas en un momento dado, una vez que esos síntomas ya son suficientemente perceptibles para que la persona los reporte. No están diseñados para detección temprana, y no pueden serlo por su naturaleza: son puntuales, dependen del autoinforme, y exigen carga sostenida del participante.

El autoinforme tiene además limitaciones específicas en el contexto premenstrual. La memoria retrospectiva de síntomas está sujeta a sesgos sistemáticos documentados: recall bias dependiente del estado emocional actual, atribución cíclica de eventos contextuales, y subreporte por estigma o normalización del malestar. El reporte intermitente reduce además la cobertura útil para validación.

> **Implicación para la validación:** si la detección fisiológica y el autoinforme no coinciden, eso no prueba que la detección falle. Puede significar que la señal fisiológica captura algo que el autoinforme no recoge, o que lo recoge con retraso, o que el reporte simplemente no está disponible ese día.

Existe un hueco clínico concreto: hay un periodo pre-sintomático observable en señales fisiológicas autonómicas, termorreguladoras, respiratorias y de sueño, que los sistemas actuales (apps de tracking calendario, escalas retrospectivas) no capturan. Durante ese periodo, una intervención preventiva (medicación intermitente, ajuste de carga, planificación) sería más eficaz que cualquier intervención reactiva. **Este proyecto apunta a ese hueco.**


## 2. Lo que se ha intentado y por qué no ha funcionado

En la última década han emergido múltiples aplicaciones comerciales de tracking del ciclo menstrual: Clue, Flo, Natural Cycles, Oura, Whoop, Apple Health. Su aproximación predominante combina tres elementos: predicción del calendario menstrual basada en histórico individual y modelos poblacionales, tracking pasivo de variables fisiológicas con análisis comparado contra rangos "normales" agregados, y logging manual de síntomas auto-reportados sin algoritmo predictivo.

Lo que estas aplicaciones no hacen es lo que define el hueco metodológico de este proyecto. No detectan idiográficamente el pródromo PMS/PMDD: no comparan a cada mujer consigo misma con baseline ajustado a su fase del ciclo, sino contra normas poblacionales o contra su propia media plana sobre ciclos completos. No anticipan temporalmente el pródromo: en el mejor de los casos avisan que la mujer está "en lútea tardía" basándose en calendario, sin detectar el cambio fisiológico que precede al síntoma. No identifican fenotipos de pródromo: no separan en qué dimensión fisiológica se manifiesta primero el cambio en cada mujer.

En la literatura científica, los estudios sobre cambios fisiológicos en el ciclo menstrual con wearables han avanzado de forma heterogénea. Alzueta et al. (2022) demostraron con Oura ring que la temperatura cutánea nocturna y la frecuencia cardíaca cambian entre las cuatro fases del ciclo (menses, ovulación, mid-luteal, late-luteal) usando *Hierarchical Linear Regression Models*. Goodale et al. (2019) con Ava bracelet aplicaron *multilevel models con random slopes y random intercepts* para detectar phase-shifts simultáneos en temperatura, frecuencia cardíaca y respiratoria. Schmalenberger et al. (2021) publicaron recomendaciones metodológicas para investigación en ciclo menstrual. El COPE study (Sichuan University, 2024) con n=193 caracterizó patrones temporales de HRV across el ciclo y su asociación con síntomas premenstruales.

Pero la revisión sistemática de Funnell et al. (2026) sobre estudios longitudinales intensivos en PMDD identifica explícitamente que la heterogeneidad metodológica importante limita la comparabilidad entre estudios, y que no existe consenso sobre el marco analítico para detección temprana del pródromo.

**El hueco que este proyecto ocupa:** ningún estudio publicado ha aplicado un marco idiográfico fase-condicional con regla de convergencia multidimensional sobre baseline individual ajustado a las cinco sub-fases del ciclo, para detección temprana del pródromo PMS/PMDD validada con verdad biológica (hormonas) y verdad clínica (síntomas). Este proyecto lo formula y lo operacionaliza.


## 3. Por qué el enfoque intra-sujeto es la respuesta

PMS/PMDD tiene expresión altamente individual. Una mujer puede tener un ciclo con síntomas severos seguido de uno con síntomas leves o ausentes. La intensidad varía entre ciclos. La dimensión fisiológica que primero se altera —HRV, temperatura, sueño, respiratoria— es distinta entre personas. Comparar a cada mujer con normas poblacionales o con su propia media plana sobre ciclos completos pierde esta variabilidad por construcción.

Si el problema es que cada persona es diferente, la solución no es entrenar modelos más sofisticados sobre datos poblacionales. Es cambiar la pregunta. En lugar de preguntar *¿se parece esta mujer al perfil de alguien con PMDD?*, la pregunta correcta es: **¿está esta mujer alejándose de su propio patrón habitual fase-condicional, en una ventana lútea tardía, en al menos dos dimensiones fisiológicas, durante al menos tres días consecutivos?**

Comparar a cada persona consigo misma en lugar de con el grupo: eso es el **enfoque intra-sujeto**. Y aplicado al ciclo menstrual exige un refinamiento: no basta con compararse con la propia media histórica, porque las señales fisiológicas tienen variación cíclica intrínseca normal. La normalidad de la HRV en fase folicular no es comparable con la normalidad en fase lútea tardía. La referencia individual debe ajustarse a la sub-fase del ciclo.

**Lo que este proyecto propone es diferente:** detectar y cuantificar la desviación respecto al baseline individual fase-condicional como señal de alerta en sí misma, sin necesidad de etiqueta clínica DSM-5. El sistema no termina en una predicción de presencia o ausencia de PMDD. Termina en una decisión: *¿cuándo es la desviación suficientemente sostenida y convergente para justificar intervención preventiva?*

La diferencia entre *"esta mujer probablemente tiene PMDD"* y *"esta mujer lleva tres días alejándose de su patrón fisiológico habitual de fase lútea tardía en dos dimensiones distintas, lo que merece atención clínica antes de que aparezcan los síntomas que probablemente reportará en los próximos días"*.


## 4. Las señales que usamos y por qué

El dataset mcPHASES (PhysioNet) recoge señales pasivas mediante Fitbit y dispositivo Mira para hormonas. Las cuatro dimensiones núcleo del proyecto son autonómica, termorregulación, respiratoria y sueño. La elección tiene respaldo empírico directo en literatura.

La **dimensión autonómica** se alimenta de RMSSD nocturno (heart rate variability) y resting heart rate. La HRV decrece entre el 10% y el 20% en fase lútea respecto a folicular en mujeres naturalmente ciclantes. La RHR aumenta ligeramente en lútea por activación simpática. Schmalenberger et al. (2021) y Alzueta et al. (2022) consolidan estos hallazgos.

La **dimensión termorregulación** se alimenta de temperatura nocturna cutánea. La progesterona produce un aumento de 0.28 a 0.56 °C tras ovulación, sostenido durante toda la fase lútea. Yu et al. (2022) y Goodale et al. (2019) confirman este efecto y demuestran su detectabilidad mediante wearable. El propio Fitbit incorpora una métrica `baseline_relative` que ya hace una primera capa de normalización individual sobre la temperatura, lo que valida empíricamente la lógica de baseline idiográfico aplicada a este sensor en producción comercial.

La **dimensión respiratoria** se alimenta de la tasa respiratoria nocturna durante sueño completo. La activación simpática lútea aumenta la frecuencia respiratoria. Goodale et al. (2019) detectó shifts simultáneos en temperatura, frecuencia cardíaca y respiratoria con Ava bracelet.

La **dimensión sueño** se alimenta del overall sleep score y restlessness. Las alteraciones del sueño son síntoma premenstrual reportado y biomarcador independiente. Baker y Driver (2007) documentan los cambios de arquitectura del sueño asociados al ciclo.

Estas cuatro dimensiones son observables de forma pasiva, continua y sin carga para la participante, y tienen significado fisiológico interpretable. Un sistema orientado a decisiones necesita poder explicar qué ha cambiado y por qué eso merece atención.

Las dimensiones secundarias (estrés, actividad) y la exploratoria (glucosa, solo en intervalo 2022) entran en interpretabilidad y análisis complementario, pero no en la regla de convergencia núcleo.


## 5. Implicación práctica y transferibilidad

Este sistema no está concebido como un modelo aislado ni como un ejercicio académico. Está diseñado como prueba de método aplicable a múltiples sectores y pacientes.

**Aplicación directa en plataformas existentes.** Las apps femtech (Clue, Flo, Natural Cycles, Oura, Whoop) ya recogen los datos pasivos necesarios. Lo que les falta es una capa idiográfica fase-condicional que permita alertar precozmente del pródromo a la usuaria. La industria farmacéutica con pipeline activo en PMDD (Vistagen PH80, Asarina Sepranolone, Lipocine LPCN 1154, Sage Therapeutics) puede usar un sistema de este tipo como endpoint exploratorio en ensayos clínicos, para identificar respondedoras y para definir la ventana de dosificación intermitente. Las CROs y plataformas DCT (Medable, Castor, ObvioHealth) pueden integrar la lógica como módulo de monitorización descentralizada con marco N-of-1 reproducible.

**Transferibilidad a otros dominios fisiológicos.** La lógica que se construye aquí —baseline individual fase-condicional, medición de desviación sostenida, convergencia multidimensional, ML supervisado sobre features SPC, interpretabilidad SHAP— es transferible a cualquier dominio donde el comportamiento fisiológico sea la señal y la detección temprana de cambio tenga valor práctico. La migraña catamenial tiene pródromo fisiológico con HRV, temperatura, sueño y respuesta electrodermal. Las crisis epilépticas tienen pródromo preictal autonómico que Yamakawa et al. (2016, 2020) ya validaron como detectable con MSPC sobre HRV en wearable. Las descompensaciones cardiovasculares, respiratorias y metabólicas en pacientes crónicos producen cambios autonómicos y respiratorios que anticipan el evento. Los brotes en enfermedades inflamatorias (Mount Sinai, 2025) son detectables hasta siete semanas antes mediante wearable.

PMS/PMDD es el dominio de aplicación de este proyecto. El sistema que se construye es más general.

## II. Problem Framing

### 1. Por qué definir el pródromo es la decisión más importante de este proyecto

Antes de construir cualquier baseline, antes de calcular cualquier z-score, antes de elegir cualquier algoritmo, hay una decisión que condiciona todo lo demás: qué entendemos exactamente por pródromo PMS/PMDD.

Esta decisión no es técnica. Es conceptual y estructural. Todo lo que viene después —las variables que se usan, cómo se transforman, cómo se combinan, cómo se interpreta el resultado— depende de cómo se defina ese punto de partida. Si la definición es débil, ambigua o equivocada, el sistema completo hereda ese problema. Puede ser técnicamente correcto y, aun así, estar resolviendo algo distinto de lo que se pretende.

El error más habitual en este tipo de proyectos sería aceptar la definición que viene dada. El dataset mcPHASES incluye autoinformes diarios de síntomas (cramps, mood swings, fatigue, sorebreasts, bloating). La tentación es utilizarlos directamente como variable objetivo. Sin embargo, esos autoinformes describen el estado sintomático ya manifiesto, no el proceso fisiológico que lo precede. Predecir el autoinforme es predecir el final del proceso, no detectar el proceso que conduce a ese estado.

La pregunta relevante no es qué etiqueta clínica está disponible, sino **qué proceso fisiológico se quiere capturar y cómo se manifiesta en los datos antes de que sea sintomáticamente evidente**.

Ese proceso no es observable de forma directa. No existe una variable única que indique pródromo. Lo disponible son señales fisiológicas parciales: HRV, temperatura, respiratoria, sueño. Cada una puede cambiar por múltiples motivos sin relación con PMS/PMDD. Una caída en HRV puede deberse a estrés laboral. Un aumento en temperatura puede deberse a habitación caliente o consumo de alcohol. Un cambio en sueño puede deberse a viaje o cafeína. Ninguna señal aislada permite inferir pródromo de forma fiable.

El problema real no es identificar cambios en variables individuales, sino **distinguir entre variabilidad cíclica normal del comportamiento fisiológico y cambios que reflejan una alteración real del funcionamiento de la persona en su fase lútea tardía**.

Para que esa distinción sea defendible, la definición de pródromo debe cumplir tres condiciones que se desarrollan en las secciones siguientes:

- **Relativa al baseline individual fase-condicional.** No interesa el nivel absoluto de una variable, sino su desviación respecto al patrón habitual de esa persona en la sub-fase del ciclo correspondiente.
- **Convergente entre dimensiones.** El pródromo no se define por un cambio en una sola dimensión fisiológica, sino por la aparición simultánea de cambios en varias.
- **Persistente.** Los cambios puntuales forman parte del comportamiento normal. Para que una señal sea interpretable, debe mantenerse en el tiempo lo suficiente como para descartar ruido.

Estas tres condiciones no eliminan completamente las fuentes de confusión, pero permiten construir una definición operativa que es robusta frente a ellas sin necesidad de modelar explícitamente cada causa alternativa.

A partir de esta definición, el pródromo deja de entenderse como una etiqueta y pasa a conceptualizarse como un **proceso dinámico fase-condicional**: un cambio sostenido en el patrón fisiológico de una persona durante la ventana lútea tardía, observable a través de la evolución conjunta de sus señales fisiológicas.

En este contexto, los autoinformes de síntomas no se utilizan para definir el pródromo. Su papel es contrastarlo. Aportan una referencia externa, sabiendo que es imperfecta. No pueden actuar como verdad, sino como un elemento más en la evaluación del sistema.


### 2. Análisis de contaminantes por dimensión fisiológica

Antes de definir qué es pródromo, hay que entender qué más puede producir los mismos patrones en los datos. Cada señal disponible está expuesta a causas alternativas que no tienen relación con PMS/PMDD. Si no se identifican y se tiene en cuenta cómo afectan a la definición, el sistema detectará cambios reales pero interpretará mal su origen.

El análisis que sigue no pretende eliminar cada contaminante uno por uno. Eso sería imposible con los datos disponibles. El objetivo es entender la naturaleza del ruido en cada dimensión, para diseñar una definición que sea robusta frente a él por construcción.

**Autonómica (HRV y resting heart rate).** El estrés psicosocial agudo —exámenes, conflictos interpersonales— reduce la HRV en pocas horas. El ejercicio reciente intenso o el sobreentrenamiento la reducen durante 24 a 48 horas. El consumo de alcohol vespertino reduce la HRV nocturna sustancialmente. La privación de sueño puntual la reduce. La cafeína tarde la reduce. La enfermedad febril aumenta la RHR y reduce la HRV. El problema conceptual es que una caída de HRV no significa automáticamente activación pródromo. Puede significar estrés, alcohol, mala noche. Esta dimensión por sí sola tiene baja especificidad.

**Termorregulación (temperatura nocturna cutánea).** La temperatura ambiental de la habitación —calefacción, ola de calor, ventilación— influye directamente. El consumo de alcohol vespertino aumenta la temperatura cutánea durante la primera mitad de la noche. La enfermedad febril o subfebril altera la señal. El ejercicio nocturno post-cena retarda el enfriamiento. Los cambios estacionales producen drift gradual. Las fluctuaciones REM/NREM provocan oscilaciones intra-noche. Fitbit ya incorpora una primera capa de defensa con la métrica `baseline_relative` que calcula desviación frente al baseline personal del propio dispositivo, absorbiendo parcialmente las fluctuaciones individuales no cíclicas. Sin embargo, no separa por sub-fase del ciclo.

**Respiratoria (tasa respiratoria nocturna).** El ejercicio cardiovascular intenso en el día previo deja una tasa respiratoria nocturna elevada. El tabaco activo produce alteración crónica. El alcohol es depresor de la tasa respiratoria pero altera la arquitectura del sueño. La apnea obstructiva del sueño no diagnosticada introduce ruido sistemático. Los procesos respiratorios infecciosos. La altitud durante un viaje produce aumento sostenido durante días.

**Sueño (overall sleep score y restlessness).** Estrés agudo, ansiedad situacional. Cafeína tarde, alcohol, comidas pesadas. Ruido ambiental, temperatura de la habitación, niños. Viaje, jet lag. Cambio de turno laboral. Procesos infecciosos. Pareja con horarios distintos. Fines de semana frente a días laborables.

**La conclusión que estructura todo lo que sigue.** Ninguna de estas dimensiones, analizada de forma independiente, permite inferir pródromo PMS/PMDD de forma fiable. Esto no invalida las señales: invalida el enfoque de mirarlas por separado. La definición de pródromo debe emerger de la combinación de señales en la ventana temporal correcta —fase lútea tardía—, diseñada para que sea difícil de explicar por causas alternativas aisladas.


### 3. Los tres criterios: persistencia, convergencia y baseline fase-condicional

Los tres criterios que estructuran la definición de pródromo en este proyecto no son convenciones arbitrarias. Son respuestas directas al problema de contaminación identificado en la sección anterior. Cada criterio elimina una clase de ruido que los otros no cubren.

**Persistencia (M = 3 días consecutivos).** Un cambio de uno o dos días en cualquier dimensión fisiológica puede ser cualquier cosa: un mal día, una mala noche, alcohol vespertino, un evento puntual. Ningún sistema que dispare alertas ante cambios de esta duración tendría utilidad práctica porque generaría falsas alarmas constantes. La persistencia exige que el cambio se mantenga al menos tres días consecutivos. Este umbral no es arbitrario. La ventana del pródromo PMS/PMDD es de aproximadamente siete días (días -7 a -1 antes de menstruación). Exigir más persistencia (M=5, M=7) reduciría la utilidad clínica anticipatoria. Exigir menos (M=1, M=2) introduciría ruido por contaminantes puntuales. M=3 mantiene cuatro o cinco días útiles de antelación dentro de la ventana lútea tardía. La persistencia sola no es suficiente: el alcohol vespertino sostenido durante tres días también puede reducir HRV. Por eso necesita combinarse con los otros dos criterios.

**Convergencia (al menos 2 de 4 dimensiones núcleo).** Si solo se ve afectada una dimensión durante tres días, todavía hay demasiadas explicaciones alternativas. Una semana de estrés laboral puede reducir HRV. Una habitación caliente puede elevar temperatura. La convergencia exige que el cambio sea simultáneo en al menos dos de las cuatro dimensiones núcleo: autonómica, termorregulación, respiratoria, sueño. Cuando dos dimensiones se alteran a la vez de forma sostenida, la probabilidad de que una única causa puntual lo explique todo disminuye significativamente. Se exige convergencia en dos de cuatro, no en las cuatro, por dos razones. Primera: el dataset tiene ruido de medición. Hay días donde el wearable no se llevó puesto, o el sensor falló, o la cobertura nocturna fue insuficiente. Exigir las cuatro dimensiones siempre significaría perder casos reales por fallos técnicos. Segunda: el pródromo no tiene una expresión universal. En algunas mujeres la primera señal es la HRV. En otras es la termorregulación. En otras el sueño. Exigir las cuatro a la vez reproduciría el mismo error de los modelos poblacionales: asumir que todas se deterioran de la misma manera.

**Desviación respecto al baseline individual fase-condicional.** Los dos criterios anteriores trabajan sobre cambios relativos, no sobre valores absolutos. Eso requiere una referencia. Esa referencia no puede ser la media del grupo porque las personas son muy distintas entre sí: hay mujeres con HRV nocturna habitual de 30 ms y otras con 90 ms, y eso es su normalidad. Tampoco puede ser la media plana del individuo sobre el ciclo entero, porque las señales fisiológicas tienen variación cíclica intrínseca normal. La HRV cae entre el 10% y el 20% en fase lútea respecto a folicular, la temperatura sube tras ovulación, la respiratoria aumenta. Si se calcula un baseline plano sobre 21 días que mezcla folicular y lútea, la varianza inflada por la variación cíclica enmascara la desviación idiográfica genuina. La referencia debe ser **fase-condicional**: para cada sub-fase del ciclo, cada mujer tiene su propio nivel habitual y su propia variabilidad habitual. La desviación se mide contra la sub-fase correspondiente al día evaluado. El detalle operativo de las cinco sub-fases se desarrolla en la sección 5.

**Los tres criterios operan juntos.** Ninguno es suficiente por sí solo. La persistencia sin convergencia detecta cambios sostenidos en una sola dimensión, que pueden tener explicaciones alternativas claras. La convergencia sin persistencia detecta coincidencias puntuales entre dimensiones, que pueden ser ruido. El baseline fase-condicional sin los otros dos criterios simplemente mide variabilidad, sin distinguir si es relevante. Es la combinación de los tres lo que produce una señal que es difícil de explicar por causas alternativas y que tiene coherencia como indicador de un posible pródromo real.


### 4. Definición formal operativa

A partir del análisis anterior, la definición operativa que guía este proyecto es la siguiente:

**Pródromo PMS/PMDD detectado = desviación con |z-score| > 1.5 sostenida durante al menos 3 días consecutivos, respecto al baseline individual fase-condicional, convergente en al menos 2 de las 4 dimensiones fisiológicas núcleo (autonómica, termorregulación, respiratoria, sueño). La detección se aplica únicamente sobre días dentro de la ventana lútea tardía (días -7 a -1 antes del próximo Menstrual), en ciclos no anovulatorios.**

Esta definición tiene cuatro propiedades que conviene hacer explícitas.

**Es relativa, no absoluta.** No hay un nivel de HRV, temperatura, respiratoria o sueño que sea en sí mismo pródromo. Lo que importa es el alejamiento de la normalidad fase-condicional propia.

**Es fisiológica, no clínica.** No depende de etiquetas diagnósticas DSM-5 ni de autoinforme. Se construye íntegramente a partir de señales pasivas longitudinales. Los autoinformes de síntomas no forman parte de la definición: son una referencia externa para contrastar el sistema, no para construirlo.

**Es dinámica, no estática.** No etiqueta a una mujer como afectada o no afectada en un momento dado. Describe un proceso fase-condicional: el cambio sostenido en el patrón fisiológico de una persona durante la ventana lútea tardía.

**Es operativa con los datos disponibles.** Cada uno de sus componentes —baseline individual fase-condicional, desviación por dimensión, persistencia temporal, convergencia entre dimensiones— puede calcularse directamente a partir del dataset mcPHASES.

Lo que esta definición no garantiza es que todos los casos detectados sean PMDD clínico en el sentido diagnóstico DSM-5. Lo que sí garantiza es que los casos detectados son difícilmente explicables por contaminantes puntuales, y que cuando no coincidan con el autoinforme, la primera hipótesis razonable no será que el sistema falla, sino que está detectando algo que el autoinforme no ha capturado todavía.

Esta definición no se traduce directamente en una clasificación binaria. El sistema no produce únicamente una etiqueta de pródromo o no pródromo, sino un **índice continuo de desviación** respecto al baseline individual fase-condicional, construido a partir de la intensidad, persistencia y convergencia de las señales. Este índice permite capturar la naturaleza gradual del proceso. No es equivalente una desviación leve en dos dimensiones durante tres días que una desviación intensa, sostenida y convergente en las cuatro dimensiones. La representación continua preserva esa información y permite analizar la evolución del estado de la persona en el tiempo.

La decisión de intervención no se deriva directamente de la definición de pródromo, sino de la aplicación de un umbral sobre el índice. Este enfoque separa la medición del fenómeno de la toma de decisión, lo que resulta más coherente con un sistema orientado a monitorización y detección temprana.

Como capa adicional para la validación clínica, se define separadamente un **ciclo con pródromo manifiesto**. Un ciclo C tiene pródromo manifiesto si, para una mujer S, la media de síntomas clave (cramps, moodswing, fatigue, sorebreasts, bloating) durante días -7 a -1 antes de menstruación es al menos un punto mayor —en escala Likert 0 a 5— que la media de los mismos síntomas durante el resto del ciclo, con cobertura mínima de cuatro de siete días. Esta definición se usa solo para validación posterior, no para entrenar el sistema.

Esta definición es el punto de partida de todo lo que sigue. A partir de aquí, el trabajo es traducirla a código.

## 5. El problema del baseline cíclico y la decisión arquitectónica

Definir cómo se construye el baseline introduce un problema conceptual que condiciona todo el sistema. Si la referencia individual no se ajusta a la sub-fase del ciclo en la que está la persona, el sistema medirá desviaciones contra una referencia mezclada y los cambios genuinos del pródromo quedarán enmascarados por la variación cíclica normal.

Este problema no tiene una solución perfecta con los datos disponibles. La elección del baseline debe equilibrar tres restricciones: la naturaleza cíclica de la señal, la cobertura limitada por sub-fase en un dataset de 90 días por sujeto, y la auditabilidad del marco como sistema decisional defendible.

El enfoque adoptado se basa en tres decisiones estructurales.

**Primero: baseline fase-condicional sobre cinco sub-fases.**

El baseline se calcula por separado para cada combinación de mujer, sub-fase y dimensión. Las cinco sub-fases son menstrual, folicular, ovulación, lútea temprana y lútea tardía. Las cuatro primeras corresponden directamente a las etiquetas `phase` del dataset mcPHASES (Menstrual, Follicular, Fertility, Luteal). La quinta —lútea tardía— se identifica operativamente como los últimos siete días con etiqueta Luteal antes del próximo Menstrual; el resto de días Luteal forman lútea temprana.

La justificación es estructural y empírica. Las señales fisiológicas tienen variación cíclica intrínseca documentada en literatura. La HRV cae entre el 10% y el 20% en lútea respecto a folicular. La temperatura nocturna sube entre 0.28 y 0.56 °C tras ovulación. La RHR aumenta ligeramente en lútea por activación simpática. La frecuencia respiratoria sube en lútea. Si el baseline mezcla estas fases, la varianza inflada por la variación cíclica enmascara la desviación idiográfica genuina del pródromo. El baseline fase-condicional separa esa variación cíclica del cambio anómalo.

Para cada par (mujer, sub-fase) y cada dimensión, el baseline se calcula como **(mediana, MAD)** en lugar de (media, desviación estándar). Con cobertura natural de 5 a 20 días por sub-fase en un sujeto-intervalo de 90 días, los estimadores robustos absorben mejor el ruido del wearable que los paramétricos. Una sola noche con artefacto de movimiento puede sesgar la media; la mediana lo absorbe.

**Segundo: baseline expandible por ciclo.**

Cada sujeto-intervalo de 90 días contiene aproximadamente tres ciclos. La estrategia de uso del baseline es expandible:

- Ciclo 1: solo construye baseline. No hay detección.
- Ciclo 2: detección activa, con baseline computado sobre los días equivalentes del ciclo 1.
- Ciclo 3: detección activa, con baseline computado sobre los días equivalentes de los ciclos 1 y 2.

Esta estructura da aproximadamente dos ventanas de detección por sujeto-intervalo. Es el precio que se paga por exigir un baseline robusto: el primer ciclo entero se dedica a calibración. La alternativa —usar baseline poblacional o asumir baseline desde el primer día— rompería la lógica idiográfica del sistema. La renuncia metodológica es explícita y aceptable.

**Tercero: filtro de ciclos anovulatorios.**

PMS/PMDD requiere ovulación. Un ciclo anovulatorio no produce el aumento de progesterona característico de la fase lútea, por lo que el pródromo asociado a esa señalización hormonal no aparece. Incluir ciclos anovulatorios en el análisis introduciría falsos negativos sistemáticos: ciclos sin pródromo porque no hay verdadera fase lútea, no porque el sistema falle.

El criterio de filtro es jerárquico. Si el ciclo tiene datos de pregnanediol glucuronido (PdG), se aplica el criterio bioquímico estándar de PdG por encima de 5 μg/mL durante al menos tres días consecutivos en fase Luteal. Si no hay datos PdG suficientes —cobertura limitada en el dataset—, se aplica como respaldo el criterio de pico LH durante fase Fertility por encima de 20 mIU/mL, valor de referencia clínico estándar. Si ninguno de los dos criterios es evaluable por falta de datos hormonales, el ciclo se mantiene en el análisis con marcador de "ovulatorio no verificable", y se reporta sensibilidad por separado.

**Riesgo principal del baseline fase-condicional: cobertura insuficiente en lútea temprana.**

La sub-fase con menos cobertura natural es lútea temprana, con mediana de unos ocho días por sujeto-intervalo. Si el criterio de cobertura mínima para considerar el baseline calculable se aplica de forma uniforme a todas las sub-fases (por ejemplo, cinco días por sub-fase), una proporción significativa de sujetos quedaría sin baseline en lútea temprana. La mitigación es definir umbrales diferenciados por sub-fase, manteniendo cinco días para las cuatro sub-fases largas y aceptando cuatro días para lútea temprana. Esto preserva la mayor parte del dataset sin comprometer la fiabilidad del estimador.

**Regla de credibilidad — Bloque 2.**

- *Decisión:* baseline fase-condicional con cinco sub-fases, estimadores robustos (mediana, MAD), expansión por ciclo, exclusión de ciclos anovulatorios mediante criterio jerárquico PdG → LH → flag.
- *Renuncia:* se renuncia al baseline plano sobre 21 días, al baseline poblacional, al baseline rolling lineal aplicable a señales no cíclicas. Se renuncia también al primer ciclo entero como ventana de detección activa.
- *Uso real:* el primer mes calibra; los meses 2 y 3 monitorizan. En despliegue clínico, esto traduce a tres meses de onboarding fisiológico antes de que el sistema empiece a alertar.
- *Trade-off:* se prioriza la solidez metodológica del baseline sobre la cantidad de ventanas de detección disponibles. Es preferible tener detección rigurosa en dos ciclos que detección dudosa en tres.

## 6. Mapeo de señales a dimensiones fisiológicas

La definición de pródromo opera sobre cuatro dimensiones fisiológicas núcleo: autonómica, termorregulación, respiratoria y sueño. Cada dimensión se alimenta de señales específicas del dataset mcPHASES. Antes de construir el baseline, es necesario hacer explícito qué señales del dataset corresponden a cada dimensión y con qué lógica se agrupan.

**Autonómica: HRV nocturna y resting heart rate.** La feature primaria de HRV es el RMSSD (root mean square of successive differences) calculado durante el sueño, agregado a nivel diario como mediana de los intervalos válidos de la noche. La feature primaria de RHR es la frecuencia cardíaca en reposo. Ambas señales reflejan tono autonómico y son sensibles a la activación simpática característica de la fase lútea. La convergencia entre HRV y RHR dentro de esta dimensión refuerza la señal: una caída de HRV acompañada de aumento de RHR es un patrón más coherente con activación simpática sostenida que cualquiera de los dos cambios aislados.

**Termorregulación: temperatura nocturna cutánea.** La feature primaria es `nightly_temperature` del archivo `computed_temperature.csv`. Fitbit incorpora además una métrica `baseline_relative` que calcula desviación frente al baseline personal del propio dispositivo. Aunque esta métrica ya hace una primera capa de normalización, no separa por sub-fase del ciclo, por lo que el sistema construye su propia normalización fase-condicional sobre la temperatura cruda y compara con la métrica de Fitbit como referencia.

**Respiratoria: frecuencia respiratoria nocturna.** La feature primaria es `full_sleep_breathing_rate` del archivo `respiratory_rate_summary.csv`, que registra la tasa respiratoria media durante el sueño completo. La activación simpática lútea aumenta la frecuencia respiratoria nocturna.

**Sueño: overall sleep score y restlessness.** Las features primarias son `overall_score` y `restlessness` del archivo `sleep_score.csv`. La señal central no es cuánto duerme la mujer en términos absolutos, sino si su patrón de sueño se vuelve más alterado o restless respecto a su baseline fase-condicional.

**Decisiones de exclusión.**

Se excluyen del núcleo las dimensiones siguientes, con justificación. **`wrist_temperature.csv`** queda fuera porque su granularidad por minuto es redundante con la agregación nocturna de `computed_temperature.csv`. **`steps.csv`** queda fuera porque la información de actividad ya está agregada en `active_minutes.csv`. **`sleep.csv`** queda fuera porque su estructura JSON anidada es redundante con `sleep_score.csv`. **`heart_rate.csv`** (1.9 GB) queda fuera porque la información autonómica relevante para pródromo está concentrada en HRV nocturno y RHR. **`height_and_weight.csv`** queda fuera por cobertura insuficiente (50% al 81% faltante).

**Dimensiones secundarias y exploratoria.**

Se mantienen como dimensiones secundarias **estrés** (`stress_score.csv`, cobertura 36 de 42 sujetos) y **actividad** (`active_minutes.csv`, cobertura 42 de 42 sujetos). Estas dimensiones no entran en la regla de convergencia núcleo, pero participan en interpretabilidad y en análisis complementario.

Se mantiene como dimensión exploratoria **metabolismo** (`glucose.csv`), con la limitación de que solo está disponible en el intervalo 2022 (42 sujeto-intervalos). El análisis de glucosa se realiza separadamente en este subconjunto.

**Nota sobre validación de cobertura por sub-fase.**

La cobertura efectiva de cada dimensión en cada sub-fase de cada sujeto-intervalo se valida empíricamente al construir el baseline. La sub-fase con menor cobertura natural es lútea temprana (mediana ocho días por sujeto-intervalo), por lo que el criterio de cobertura mínima se ajusta a esta realidad: cinco días para sub-fases largas, cuatro días para lútea temprana.

## 7. Regla de credibilidad aplicada al Bloque

Cada bloque de este proyecto cierra con cuatro preguntas que distinguen un ejercicio académico de un trabajo con criterio aplicado.

**Decisión.** Rechazar el autoinforme de síntomas como variable objetivo y construir una definición fisiológica operativa propia del pródromo PMS/PMDD, basada en desviación intra-sujeto fase-condicional sostenida y convergente en al menos dos de las cuatro dimensiones núcleo durante al menos tres días consecutivos en la ventana lútea tardía, en ciclos no anovulatorios.

**Renuncia.** Se renuncia a la comodidad de usar las puntuaciones de síntomas auto-reportados como target supervisado directo. Se renuncia a la comparabilidad directa con literatura del nicho específico que usa hierarchical regression o multilevel models como marco principal. Se renuncia al baseline plano y al rolling lineal —que sí son válidos en proyectos sobre señales no cíclicas como GLOBEM, pero introducirían varianza inflada en este caso—. Se renuncia a los marcos modernos más sofisticados como similarity-based modeling con digital twin o pattern mixture models por razones técnicas concretas: el primero requiere data densa por sujeto que no está disponible con tres ciclos de 90 días; el segundo está optimizado para data de baja frecuencia con cambio post-onset, no para pródromo cíclico recurrente.

**Uso real.** En una plataforma femtech o en un estudio clínico descentralizado, el primer mes de cada usuaria es onboarding fisiológico: el sistema acumula datos para construir el primer baseline fase-condicional. A partir del segundo mes, el sistema evalúa día a día —dentro de la ventana lútea tardía de cada ciclo— si las dimensiones fisiológicas convergen en desviación sostenida respecto a la propia normalidad de la usuaria. Si lo hacen, se dispara una alerta orientada a decisión de intervención preventiva: dosificación intermitente de medicación, ajuste de carga laboral, planificación. El sistema no diagnostica PMDD ni reemplaza evaluación clínica. Sirve para anticipar el momento en que el pródromo está empezando, antes de que los síntomas sean manifiestos.

**Trade-off.** Se gana la capacidad de detectar el pródromo antes de que sea sintomáticamente evidente, sin depender del autoinforme y sin asumir homogeneidad inter-sujeto. Se pierde la certeza de estar midiendo exactamente PMDD clínico en el sentido diagnóstico DSM-5. El sistema detecta cambio fisiológico sostenido fase-condicional, no diagnóstico. Esa diferencia es deliberada y está documentada. Se acepta también el riesgo de que la detección fisiológica no supere al calendario menstrual como baseline trivial. Si eso ocurre en la evaluación, es hallazgo válido y honesto que se documenta. El proyecto no depende de obtener un resultado positivo: depende del rigor metodológico del trabajo.

## 8. Marco científico y técnica estadística

Este proyecto se enmarca en una tradición metodológica concreta que conviene hacer explícita.

**Marco científico: N-of-1 longitudinal observational design.**

N-of-1 es un diseño científico en el que la unidad de análisis es el individuo, estudiado con medidas repetidas en el tiempo. En lugar de comparar grupos para sacar conclusiones poblacionales, se estudia cómo cambia cada persona consigo misma. Tiene décadas de literatura en psicología clínica, en single-case experimental designs, y en medicina personalizada. La pregunta que define el proyecto, *¿se está esta mujer alejándose de su propio patrón fase-condicional?*, es N-of-1 puro. No es enfoque poblacional disfrazado.

En LUTEA, la unidad de análisis es el **sujeto-intervalo**: cada uno de los 62 sujeto-intervalos del dataset mcPHASES (42 en 2022 más 20 en 2024) se trata como una unidad N-of-1 independiente. Para los 20 sujetos con datos en ambos intervalos, separados dos años, se tratan como dos unidades independientes, no como seguimiento longitudinal continuo.

**Técnica estadística: Statistical Process Control con extensiones modernas.**

La lógica operativa del sistema —establecer baseline, medir desviación, distinguir señal de ruido, detectar cambio sostenido, decidir intervención— coincide punto por punto con el SPC desarrollado por Walter Shewhart en los años 20 del siglo pasado. SPC es marco consolidado en healthcare con literatura sistemática reciente que documenta su aplicación en monitorización individual de pacientes con asma, hipertensión y otras condiciones crónicas (Pérez-Benítez et al., 2023; Waqas et al., 2024). La diferencia respecto al SPC industrial es que las señales fisiológicas femeninas tienen propiedades que las industriales no tienen: ciclicidad de aproximadamente 28 días, autocorrelación, contaminantes contextuales (alcohol, ejercicio, estrés), missing data, y series no estacionarias.

Por eso no se aplica Shewhart "de manual"; se aplican adaptaciones modernas. Baseline individual fase-condicional sobre cinco sub-fases en lugar de baseline plano poblacional. Z-score intra-sujeto contra el baseline de la sub-fase correspondiente al día evaluado. Run rules de persistencia (M=3) y convergencia multidimensional (2 de 4). Estimadores robustos (mediana, MAD) en lugar de paramétricos. Métodos modernos como CUSUM y EWMA en el Bloque 7 para captura de derivas lentas. Change point detection (BOCPD, PELT) como ventana adicional. Estos métodos no dependen del baseline fase-condicional y detectan cambios estructurales en la serie completa, lo que cubre el caso del pródromo gradual que el z-score puntual contra baseline podría no señalar.

**Precedente metodológico directo del paradigma SPC + HRV + wearable + pródromo.**

Yamakawa et al. publicaron en *IEEE Transactions on Biomedical Engineering* (2016) y en *Sensors* MDPI (2020) un sistema de predicción de crisis epilépticas mediante análisis HRV integrado con anomaly monitoring usando **Multivariate Statistical Process Control** (MSPC). El sistema wearable monitoriza ocho features de HRV en tiempo real con MSPC. El paradigma es estructuralmente idéntico al de este proyecto: pródromo fisiológico (preictal en epilepsia, lútea tardía en LUTEA), HRV como ventana al sistema nervioso autónomo, anomaly monitoring multivariado, dispositivo wearable. Esto constituye precedente publicado y citable que valida la elección de SPC para señales fisiológicas pasivas con pródromo.

**Alternativas modernas evaluadas y descartadas.**

Antes de elegir SPC se evaluaron explícitamente otras tradiciones metodológicas. *Similarity-Based Modeling con physiological digital twin* (Steinhubl et al., 2024-2025): estado del arte 2025 para detección de cambios fisiológicos sutiles, pero requiere data densa por sujeto que no está disponible con tres ciclos de 90 días. *Pattern Mixture Models* (Han et al., 2020): superan a SREM y ROCA en detección temprana de cáncer de ovario con CA125 anual, pero están optimizados para baja frecuencia de medida con cambio post-onset, no para pródromo cíclico recurrente. *Multilevel models con random slopes y random intercepts* (Goodale et al., 2019, con Ava bracelet): captura formalmente variabilidad inter-cíclica e inter-sujeto, pero pierde la interpretabilidad y auditabilidad de SPC, que son ventajas reales en el contexto regulatorio que requiere el caso de uso. *Circular statistics + Rayleigh test* (Karoly et al., 2018): específicas para datos de naturaleza periódica, pero capturan periodicidad a nivel de ciclo completo, no son adecuadas para detección dentro de una sub-fase específica con regla de convergencia.

La elección de SPC se justifica por cinco razones concretas: precedente publicado directo (Yamakawa); marco consolidado en healthcare con literatura sistemática reciente; adecuación a un dataset de 62 sujeto-intervalos sin requisito de data densa por sujeto; interpretabilidad y auditabilidad nativas, compatibles con regulación; coherencia metodológica con el proyecto gemelo (GLOBEM) en el portfolio del autor.

**La integración de las dos capas.**

N-of-1 responde a *qué unidad analítica estudio*: el individuo. SPC responde a *cómo detecto cambios dentro de esa serie temporal*: límites, z-scores, run rules, CUSUM, EWMA. No son alternativas: son capas complementarias. El proyecto se describe profesionalmente como *longitudinal statistical monitoring based on patient-specific phase-conditional baseline deviation and change detection*, lenguaje que conecta directamente con el sector de digital health y biomarcadores digitales.

## III. Construcción del Baseline Individual

### 1. Qué es el baseline y por qué es la primera pieza técnica del sistema

Todo lo que sigue en este proyecto —medir desviación, detectar persistencia, evaluar convergencia— requiere una referencia. Esa referencia es el baseline: el patrón habitual de funcionamiento fisiológico de cada mujer, estimado a partir de sus propios datos y ajustado a la sub-fase del ciclo en la que está cada día evaluado.

El baseline no es un número. Para cada sujeto-intervalo, cada dimensión fisiológica y cada una de las cinco sub-fases del ciclo, el baseline es un par de estadísticos: la **mediana** y la **desviación absoluta de la mediana (MAD)** calculadas sobre los días pasados del propio sujeto-intervalo que pertenecen a esa sub-fase. La mediana captura el nivel habitual de la mujer en esa sub-fase. El MAD captura cuánta variabilidad es normal para ella en ese momento del ciclo. Ambos son necesarios: sin la mediana no hay referencia; sin el MAD no se puede distinguir un cambio real de la fluctuación cíclica habitual.

La referencia individual se construye fase-condicional en lugar de plana sobre el ciclo completo. Esta decisión, central en el diseño del baseline, adapta la idea de "patrón habitual" a la realidad de las señales fisiológicas femeninas: tienen variación cíclica intrínseca documentada en literatura. La HRV cae entre el 10% y el 20% en lútea respecto a folicular. La temperatura nocturna sube entre 0.28 y 0.56 °C tras ovulación. Si el baseline mezcla estas fases, la varianza inflada por la variación cíclica enmascara la desviación idiográfica genuina del pródromo.

La referencia se actualiza por ciclo en lugar de fijarse al inicio del registro. El baseline es expandible: el ciclo 1 calibra y no produce detección; el ciclo 2 detecta usando el baseline construido sobre los días equivalentes del ciclo 1; el ciclo 3 detecta usando los ciclos 1 y 2 acumulados. Cada mujer construye su propia escala fase-condicional, y esa escala crece con su histórico personal de ciclos completados.

Esta decisión tiene una implicación directa sobre cómo se medirá el pródromo más adelante. El z-score intra-sujeto fase-condicional —la métrica central del sistema— se calcula como la distancia entre el valor observado el día d y la mediana del baseline correspondiente a la sub-fase de d, dividida por el MAD escalado de esa misma sub-fase. Una caída de RMSSD de 15 ms puede ser señal de alarma para una mujer cuyo RMSSD habitual en lútea tardía es 60 ms con MAD pequeño, y ruido para otra que fluctúa habitualmente entre 30 y 80 ms en la misma sub-fase.

En lenguaje técnico, este baseline establece el "proceso bajo control" individual fase-condicional para cada sujeto-intervalo en cada momento del ciclo: la lógica de Statistical Process Control aplicada a monitorización N-of-1 sobre series con periodicidad cíclica intrínseca, donde cada mujer tiene su propia línea central y su propia franja de variabilidad normal por sub-fase, ambas actualizadas con su histórico de ciclos completados.

### 2. Qué señales entran en el baseline

Cada dimensión fisiológica núcleo se representa con una **feature primaria**: la señal más directa, interpretable y con mejor cobertura empírica dentro de esa dimensión.

**Autonómica (sistema nervioso autónomo)**
- HRV: RMSSD nocturno, agregado como mediana de los intervalos de 5 minutos válidos de la noche (feature primaria).
- RHR: frecuencia cardíaca en reposo (feature primaria complementaria dentro de la misma dimensión).

**Termorregulación (regulación térmica nocturna)**
- Temperatura cutánea nocturna (feature primaria).

**Respiratoria (control autonómico de la respiración)**
- Frecuencia respiratoria durante sueño completo (feature primaria).

**Sueño (arquitectura y calidad del descanso)**
- Overall sleep score (feature primaria).
- Restlessness (feature primaria complementaria dentro de la misma dimensión).

El resto de variables disponibles en el dataset mcPHASES (estrés, actividad, glucosa) entran como dimensiones secundarias o exploratorias en fases posteriores del proyecto. Para el baseline, se busca una señal limpia y estable por dimensión núcleo: mezclar demasiadas features aquí introduce ruido y dificulta la interpretación cuando algo se desvía.

### 3. Criterio de cobertura y validez por sub-fase

El baseline fase-condicional de un sujeto-intervalo en una sub-fase concreta requiere que esa sub-fase tenga suficientes días con medida válida para que la mediana y el MAD sean estimaciones fiables. El criterio operativo es:

**Cobertura mínima diferenciada por sub-fase.** Al menos 5 días con valor no nulo en sub-fases largas (Menstrual, Folicular, Fertility, Lútea tardía); al menos 4 días en Lútea temprana, que tiene una cobertura natural más baja —mediana de 8 días por sujeto-intervalo— por la duración estructural más corta de esa sub-fase. Si una sub-fase no cumple esta condición en una dimensión concreta, el baseline para ese par (sub-fase, dimensión) se marca como no calculable y los días de esa sub-fase en ese sujeto-intervalo quedan no evaluables en esa dimensión.

**Política de gaps.** Los días sin medida válida en una dimensión se mantienen como `NaN` y no se imputan. La mediana y el MAD se calculan sobre los días no nulos de la sub-fase. La regla de convergencia 2 de 4 que se aplica en la fase de detección permite que una dimensión `NaN` un día concreto no cuente, sin invalidar la evaluación de las otras tres dimensiones.

**Filtro biológico previo: ciclos anovulatorios.** PMS/PMDD requiere ovulación. Un ciclo anovulatorio no produce el aumento de progesterona característico de la fase lútea, por lo que el pródromo asociado a esa señalización hormonal no aparece. Antes de entrar al baseline, cada ciclo se evalúa con un criterio jerárquico: primero PdG > 5 µg/mL durante al menos 3 días consecutivos en Luteal cuando hay datos suficientes; en su defecto, LH max > 20 mIU/mL en Fertility; si ninguno es evaluable, el ciclo se mantiene en el análisis con marcador "ovulatorio no verificable" y se reporta sensibilidad por separado. Los ciclos clasificados como anovulatorios se excluyen del análisis principal de pródromo.

**Calibración expandible por ciclo.** El primer ciclo completo de cada sujeto-intervalo se dedica íntegramente a construir el baseline. No produce detección. A partir del ciclo 2, el sistema evalúa día a día los días que pertenecen a la ventana lútea tardía contra el baseline construido con los ciclos pasados del propio sujeto-intervalo.

No se aplica filtro de estabilidad inter-ciclo. La razón es estructural: el baseline es expandible, y la variabilidad legítima entre ciclos de la misma mujer queda absorbida en el MAD fase-condicional, que es precisamente el estimador de variabilidad normal de esa sub-fase para ella.

No se pone umbral de varianza absoluta. Hay mujeres con MAD naturalmente alto en alguna dimensión y eso es su variabilidad habitual. Excluirlas por ser variables sería introducir un sesgo que el enfoque intra-sujeto fase-condicional está diseñado precisamente para evitar.

### 4. Salida del módulo de baseline

La aplicación del criterio de cobertura por sub-fase y del filtro anovulatorio produce, para cada sujeto-intervalo, una clasificación de qué dimensiones y qué sub-fases tienen baseline calculable y, derivada de ella, una clasificación del propio sujeto-intervalo en función de la calidad del baseline construido:

- **Sujeto-intervalo completo.** Las cuatro dimensiones núcleo tienen baseline calculable en al menos cuatro de las cinco sub-fases, incluyendo obligatoriamente Lútea tardía y Folicular como referencias críticas. El sistema puede medir desviación fase-condicional con la lógica completa.
- **Sujeto-intervalo parcial.** Tres de las cuatro dimensiones núcleo cumplen el criterio anterior, o las cuatro pero solo en tres sub-fases. Se mantiene en análisis usando solo las dimensiones válidas; la regla de convergencia opera sobre el subconjunto disponible.
- **Sujeto-intervalo insuficiente.** Menos de tres dimensiones núcleo con baseline calculable, o falta de Lútea tardía. Excluido del análisis principal y reportado como limitación.

Esta clasificación no es una etiqueta estática global. Para los sujeto-intervalos completos y parciales, la evaluabilidad opera además a nivel de día: un día concreto de la ventana lútea tardía es evaluable en una dimensión solo si esa dimensión tiene medida válida ese día y el baseline de esa (sub-fase, dimensión) es calculable. La regla de convergencia 2 de 4 entre dimensiones se aplica solo a los días en que el sujeto-intervalo tiene al menos dos dimensiones evaluables.

La dimensión autonómica tiene dos features primarias (HRV y RHR). Se considera evaluable un día concreto si al menos una de las dos tiene baseline fase-condicional calculable y medida válida ese día. La dimensión sueño sigue la misma lógica con overall sleep score y restlessness.

Esta lógica no es un filtro de exclusión muestral. Es una funcionalidad del sistema que reduce la fatiga de alerta por baselines no fiables y documenta explícitamente sobre qué ciclos, qué sub-fases y qué dimensiones de cada sujeto-intervalo opera el sistema y sobre cuáles no.

### Regla de credibilidad (Bloque 2)

**Decisión:** baseline individual fase-condicional sobre cinco sub-fases del ciclo (Menstrual, Folicular, Fertility, Lútea temprana, Lútea tardía), definido como (mediana, MAD) por sujeto-intervalo, dimensión y sub-fase, calculado sobre los días pasados del propio sujeto-intervalo que pertenecen a cada sub-fase, con criterios de cobertura mínima diferenciados por sub-fase, filtro anovulatorio jerárquico previo (PdG → LH → flag), una feature primaria por dimensión núcleo, y expansión por ciclo (ciclo 1 calibra, ciclos posteriores detectan).

**Renuncia:** se renuncia al baseline plano sobre el ciclo completo, al baseline rolling lineal aplicable a señales no cíclicas y al baseline poblacional. Se renuncia a la comodidad de un único par (mediana, MAD) por sujeto-intervalo: el baseline fase-condicional produce hasta veinte pares por sujeto-intervalo (cinco sub-fases × cuatro dimensiones), lo que aumenta la complejidad de auditoría y exige documentar la referencia exacta usada en cada evaluación. Se renuncia también al primer ciclo completo de cada sujeto-intervalo como ventana de detección activa: se dedica íntegramente a calibración. En el dataset, esto implica perder cuatro sujeto-intervalos (6.5%) que solo tienen un ciclo completo —calibran y se acaba el registro— y los 283 días Lútea sin próximo Menstrual capturado (5% del dataset).

**Uso real:** en una plataforma femtech o en un estudio clínico descentralizado, el primer mes aproximado de cada usuaria —su primer ciclo completo— es onboarding fisiológico fase-condicional: no se disparan alertas, el sistema acumula el histórico necesario para construir el primer baseline por sub-fase. A partir del segundo ciclo, el sistema evalúa día a día qué dimensiones tienen baseline calculable y medida válida ese día dentro de la ventana lútea tardía, y aplica la regla de convergencia 2 de 4 sobre los días evaluables. En el dataset mcPHASES, esta arquitectura produce 114 ventanas de detección reales en 58 sujeto-intervalos.

**Trade-off:** solidez del baseline frente a cantidad de ventanas de detección. El baseline fase-condicional sobre cinco sub-fases es necesario porque las señales fisiológicas femeninas tienen variación cíclica intrínseca documentada: sin separar fases, la varianza inflada por esa variación enmascara la desviación idiográfica genuina del pródromo. El precio de esa separación es que el primer ciclo completo de cada sujeto-intervalo no produce detección y que cuatro sujeto-intervalos del dataset, los que solo tienen un ciclo completo, quedan sin ventana de detección activa. Se prioriza la solidez metodológica del baseline y la auditabilidad ante regulador clínico sobre la capacidad predictiva pura. Es preferible tener detección rigurosa fase-condicional en dos ciclos por sujeto-intervalo que detección dudosa contra una referencia plana en tres ciclos.

A partir de aquí comienza la implementación del sistema.
